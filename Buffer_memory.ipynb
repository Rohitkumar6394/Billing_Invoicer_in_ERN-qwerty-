```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    device_map="auto"
)

print("Model Loaded Successfully")
```

```python
MEMORY_FILE = "memory.txt"

# Create file if not exists
open(MEMORY_FILE, "a").close()

def load_memory():
    with open(MEMORY_FILE, "r", encoding="utf-8") as f:
        return f.read()

def save_memory(text):
    with open(MEMORY_FILE, "a", encoding="utf-8") as f:
        f.write(text + "\n")
```

```python
def chat_with_agent(user_input):

    memory_str = load_memory()
    messages = []

    # Parse memory_str into messages for chat template
    lines = [line.strip() for line in memory_str.split('\n') if line.strip()]
    for line in lines:
        if line.startswith("User: "):
            messages.append({"role": "user", "content": line[len("User: "):]})
        elif line.startswith("Assistant: "):
            messages.append({"role": "assistant", "content": line[len("Assistant: "):]})

    # Add current user input
    messages.append({"role": "user", "content": user_input})

    # Apply chat template using the generator's tokenizer
    prompt = generator.tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

    response = generator(
        prompt,
        max_new_tokens=80,
        do_sample=True,
        temperature=0.5,
        return_full_text=False
    )

    answer = response[0]["generated_text"].strip()

    # Save memory for both user input and assistant's answer
    save_memory(f"User: {user_input}")
    save_memory(f"Assistant: {answer}")

    return answer
```

```python
with open(MEMORY_FILE, "w", encoding="utf-8") as f:
    f.write("")
print("Memory cleared successfully.")
```

```python
while True:

    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Agent Stopped")
        break

    response = chat_with_agent(user_input)

    print("Agent:", response)
```
